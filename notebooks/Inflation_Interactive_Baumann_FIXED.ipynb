## Figure 1 · Δεν είναι όλες οι διαστολές ίδιες

Για τη σύγκριση θέτουμε $a(t_i)=1$, $\tau=H_i(t-t_i)$ και ίδιο αρχικό $H_i$ για τα τρία διαφορετικά, **επίπεδα** Σύμπαντα:

$$
\boxed{a_r(\tau)=(1+2\tau)^{1/2}},\quad
\boxed{a_m(\tau)=(1+\tfrac32\tau)^{2/3}},\quad
\boxed{a_\Lambda(\tau)=e^{\tau}}.
$$

**Πείραμα:** αύξησε τη χρονική διάρκεια. Η ύλη και η ακτινοβολία συνεχίζουν να διαστέλλονται αλλά **επιβραδυνόμενα** ($\ddot a<0$), ενώ η de Sitter περίπτωση επιταχύνεται ($\ddot a>0$).

```python
def compare_expansion(tau_max=3.5, log_y=False):
    tau = np.linspace(0, tau_max, 280)
    rad = np.sqrt(1 + 2*tau)
    mat = (1 + 1.5*tau)**(2/3)
    inf = np.exp(tau)
    fig, ax = plt.subplots()
    ax.plot(tau, rad, lw=2.6, label=r"Radiation  $w=1/3$")
    ax.plot(tau, mat, lw=2.6, label=r"Matter  $w=0$")
    ax.plot(tau, inf, lw=2.6, label=r"de Sitter  $w=-1$")
    ax.set(xlabel=r"$\tau=H_i(t-t_i)$", ylabel=r"$a(\tau)/a_i$", title="Three expansion histories (same initial Hubble rate)")
    if log_y: ax.set_yscale('log')
    ax.legend(frameon=False)
    plt.show()

w1 = widgets.interactive(compare_expansion,
    tau_max=widgets.FloatSlider(value=3.5,min=0.5,max=5,step=0.25,description="Duration",continuous_update=False),
    log_y=widgets.Checkbox(value=False, description="Logarithmic y-axis"))
display(w1)
```

## Figure 2 · Συνεκτεινόμενη ακτίνα Hubble και $w$

$$
\boxed{r_H(a)=(aH)^{-1}},\qquad
\boxed{\frac{r_H(N)}{r_{H,i}}=\exp\!\left[\frac{1+3w}{2}N\right]},
\quad N\equiv\ln(a/a_i).
$$

Για ένα επίπεδο Σύμπαν με ένα ρευστό σταθερού $w$, **πληθωρισμός** σημαίνει $w<-1/3$. Στο διάγραμμα δείχνουμε $\log_{10}[r_H(N)/r_{H,i}]$, ώστε να φαίνονται καθαρά οι εκθετικές μεταβολές.

**Πείραμα:** μετακίνησε το $w$ πάνω και κάτω από το $-1/3$. Προσοχή: *Hubble radius* $\neq$ *particle horizon*.

```python
def comoving_hubble(w=-0.8, Nmax=6):
    N = np.linspace(0, Nmax, 240)
    fig, ax = plt.subplots()
    for val, name, style in [(-1.0,"de Sitter",'--'), (0.0,"Matter",'--'),(1/3,"Radiation",'--'),(w,f"Selected w={w:+.2f}",'-')]:
        logr = (1+3*val)*N/(2*np.log(10))
        ax.plot(N, logr, ls=style, lw=3 if style=='-' else 1.5,label=name)
    ax.axhline(0, color='gray', lw=.8)
    ax.set(xlabel=r"E-folds  $N=\ln(a/a_i)$", ylabel=r"$\log_{10}(r_H/r_{H,i})$", title="Comoving Hubble radius: shrinking or growing?")
    ax.legend(frameon=False, ncol=2)
    plt.show()
    epsilon = 1.5*(1+w)
    print(f"ε_H = {epsilon:.3f} | " + ("Accelerated expansion (inflation)" if epsilon < 1 else "No accelerated expansion" if epsilon > 1 else "Boundary ε_H=1"))

display(widgets.interactive(comoving_hubble,
    w=widgets.FloatSlider(value=-.8,min=-1,max=1,step=.05,description="w",continuous_update=False),
    Nmax=widgets.IntSlider(value=6,min=2,max=10,step=1,description="Max N",continuous_update=False)))
```

## Figure 3 · Flatness Problem: καταστολή της καμπυλότητας

Για αρχικό $\Omega_{k,i}>0$ και ένα ρευστό σταθερού $w$, η **ακριβής** λύση της εξίσωσης Friedmann δίνει (Baumann 4.25):

$$
\boxed{\Omega_k(N)=\frac{\Omega_{k,i}e^{(1+3w)N}}{1-\Omega_{k,i}+\Omega_{k,i}e^{(1+3w)N}}}.
$$

Κατά τον πληθωρισμό τύπου de Sitter ($w=-1$) η $\lvert\Omega_k\rvert$ μικραίνει περίπου ως $e^{-2N}$, **όταν η καμπυλότητα είναι ήδη μικρή**. Κατά την ακτινοκρατούμενη εποχή $w=1/3$, μια αρχικά μικρή απόκλιση μεγαλώνει.

**Πείραμα:** άλλαξε τον αρχικό βαθμό καμπυλότητας και τη διάρκεια $N$.

```python
def curvature_history(log10_initial=-2.0, Nmax=10):
    oki = 10**log10_initial
    N = np.linspace(0,Nmax,260)
    def ok(N, w):
        # algebraically stable expression for Ωki in (0,1)
        return 1/(1+(1-oki)/oki*np.exp(-(1+3*w)*N))
    ok_inf = ok(N,-1)
    ok_rad = ok(N,1/3)
    fig,ax=plt.subplots()
    ax.plot(N,np.log10(ok_inf),label='Inflation (w = -1)',lw=2.6)
    ax.plot(N,np.log10(ok_rad),label='Radiation (w = 1/3)',lw=2.6)
    ax.set(xlabel='E-folds N',ylabel=r'$\log_{10} |\Omega_k|$',title='Evolution of curvature in two idealized models')
    ax.legend(frameon=False)
    plt.show()
    print(f"Initial Ωk = {oki:.4g};  at N={Nmax}: Ωk(inflation)={ok_inf[-1]:.3e}, Ωk(radiation)={ok_rad[-1]:.3e}")

display(widgets.interactive(curvature_history,
    log10_initial=widgets.FloatSlider(value=-2,min=-4,max=-1,step=.5,description='log₁₀ Ωki',continuous_update=False),
    Nmax=widgets.IntSlider(value=10,min=2,max=40,step=2,description='Max N',continuous_update=False)))
```

## Figure 4 · Inflaton: δυναμικό, επιλεγμένη θέση και slow-roll παράμετροι

Για ένα ομογενές, κανονικά κανονικοποιημένο βαθμωτό πεδίο:

$$
\boxed{\rho_\phi=\tfrac12\dot\phi^2+V(\phi)},\qquad
\boxed{p_\phi=\tfrac12\dot\phi^2-V(\phi)}.
$$
$$
\boxed{H^2\simeq\frac{V}{3M_{\rm Pl}^2}},\qquad
\boxed{3H\dot\phi\simeq -V_{,\phi}},\qquad
\boxed{\epsilon_V=\frac{M_{\rm Pl}^2}{2}\left(\frac{V_{,\phi}}V\right)^2}.
$$

Σχεδιάζουμε **δύο διδακτικά δυναμικά** (με $M_{\rm Pl}=1$ και αυθαίρετη κατακόρυφη κανονικοποίηση):

1. **Quadratic**: $V\propto \phi^2$ με $\epsilon_V=2/\phi^2$.
2. **Plateau / Starobinsky-like**: $V\propto(1-e^{-\sqrt{2/3}\,\phi})^2$ με $\epsilon_V=\tfrac43[e^{-\sqrt{2/3}\phi}/(1-e^{-\sqrt{2/3}\phi})]^2$.

**Πείραμα:** μετακίνησε το $\phi$. Για αρκετά μεγάλες τιμές, τα δυναμικά μπορούν να έχουν $\epsilon_V\ll1$. Το $\epsilon_V\approx1$ είναι ενδεικτικό τέλος slow roll — η ακριβής συνθήκη τέλους επιτάχυνσης είναι $\epsilon_H=1$. Το τετραγωνικό δυναμικό είναι εδώ **διδακτικό παράδειγμα**, όχι ισχυρισμός ότι συμφωνεί με όλα τα σημερινά παρατηρησιακά δεδομένα.

```python
def inflaton_potential(model='Plateau', phi=4.0):
    xx=np.linspace(.1,7.5,350)
    b=np.sqrt(2/3)
    if model=='Quadratic':
        vv=lambda x: 0.5*x*x
        eps=lambda x: 2/x**2
        eta=lambda x: 2/x**2
    else:
        vv=lambda x: (1-np.exp(-b*x))**2
        eps=lambda x: (4/3)*(np.exp(-b*x)/(1-np.exp(-b*x)))**2
        eta=lambda x: (4/3)*np.exp(-b*x)*(2*np.exp(-b*x)-1)/(1-np.exp(-b*x))**2
    fig,ax=plt.subplots()
    ax.plot(xx,vv(xx),lw=2.6,label=r'$V(\phi)$')
    ax.scatter([phi],[vv(phi)],s=85,zorder=5,label=f'φ/Mpl = {phi:.2f}')
    ax.axvline(phi,ls=':',color='gray')
    ax.set(xlabel=r'$\phi/M_{\mathrm{Pl}}$',ylabel=r'$V / V_0$ (illustrative normalization)',title=model+' inflaton potential')
    ax.legend(frameon=False)
    plt.show()
    ev=float(eps(phi)); et=float(eta(phi));
    print(f"ε_V={ev:.4f}, η_V={et:.4f}  |  " + ("slow-roll ε condition satisfied" if ev<0.1 else "slow-roll ε condition not strongly satisfied"))
    print('Για πλήρες slow roll απαιτούνται επιπλέον κατάλληλες συνθήκες για η_V και τη δυναμική του πεδίου.')

display(widgets.interactive(inflaton_potential,
    model=widgets.ToggleButtons(options=['Plateau','Quadratic'],description='Potential'),
    phi=widgets.FloatSlider(value=4,min=.5,max=7,step=.1,description='φ/Mpl',continuous_update=False)))
```

## Figure 5 · E-fold calculator

$$
\boxed{N=\ln(a_f/a_i)},\qquad
\boxed{a_f/a_i=e^N}.
$$

Για σχεδόν σταθερό $H$ (ιδανικό de Sitter):

$$
\boxed{r_{H,f}/r_{H,i}=e^{-N}},\qquad
\boxed{|\Omega_{k,f}|/|\Omega_{k,i}|\simeq e^{-2N}}
$$

Η τελευταία σχέση προϋποθέτει αμελητέα καμπυλότητα και σχεδόν σταθερό $H$. Οι ποσότητες στο διάγραμμα δίνονται ως $\log_{10}$ για να φαίνονται μεταβολές πολλών τάξεων μεγέθους.

```python
def efold_lab(N=60):
    Ns=np.linspace(0,70,220)
    fig,ax=plt.subplots()
    ax.plot(Ns,Ns/np.log(10),label=r'$\log_{10}(a/a_i)$',lw=2.4)
    ax.plot(Ns,-Ns/np.log(10),label=r'$\log_{10}(r_H/r_{H,i})$',lw=2.4)
    ax.plot(Ns,-2*Ns/np.log(10),label=r'$\log_{10}(|\Omega_k|/|\Omega_{k,i}|)$',lw=2.4)
    ax.axvline(N,ls=':',color='black')
    ax.set(xlabel='E-folds N',ylabel='Log10 of ratio',title='The power of exponential expansion')
    ax.legend(frameon=False)
    plt.show()
    print(f"N={N}: scale factor ×{np.exp(N):.3e} | comoving Hubble radius ×{np.exp(-N):.3e} | curvature (approx.) ×{np.exp(-2*N):.3e}")

display(widgets.interactive(efold_lab,
    N=widgets.IntSlider(value=60,min=0,max=70,step=1,description='E-folds N',continuous_update=False)))
```
